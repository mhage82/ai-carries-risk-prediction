```python
ctc_cols = [c for c in df.columns if c.endswith('CTC')]
```

```python
has_untreated_coronal = (
    df[ctc_cols]
    .apply(lambda row: (row == b'U').any(), axis=1)
)
```

```python
has_untreated_coronal
```

```python
has_root_caries = df["OHXRCAR"] == 1

df["has_caries"] = (has_root_caries | has_untreated_coronal).astype(int)
```
